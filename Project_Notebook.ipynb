# Email Spam Classifier
Classifying emails as spam or ham using Naive Bayes.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

data = {
    'text': [
        'Congratulations! You have won a free lottery of $1000.',
        'Hey, are we still meeting for lunch today?',
        'URGENT: Your account has been compromised.',
        'The meeting notes from yesterday are attached.',
        'Get a 50% discount on all luxury watches.',
        'Can you please send me the report?'
    ],
    'label': [1, 0, 1, 0, 1, 0]
}
df = pd.DataFrame(data)
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.2, random_state=42)

vectorizer = TfidfVectorizer(stop_words='english')
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

y_pred = model.predict(X_test_tfidf)
print(f'Accuracy: {accuracy_score(y_test, y_pred)}')
```
